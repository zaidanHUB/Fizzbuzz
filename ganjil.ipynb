```python
def menu():
    print("1. A pangkat B")
    print("2. Hitung 1 -2/3 + 5/8 - 13/21 +")
    print("0. keluar")

def pangkat():
    a = int(input("masukan suatu bilangan bulat : "))
    b = int(input("masukan pangkat yang diinginkan : "))
    for i in range(1, b+1):
        print("hasil", a, "pangkat", i, "adalah", a**i)

def deret():
    n = int(input("Masukkan jumlah N: "))

    num1 = 1
    num2 = 2
    den1 = 1
    den2 = 3

    total = 0
    for i in range(1, n+1):
        if i == 1:
            p = 1
            q = 1
        elif i == 2:
            p = num2
            q = den2
        else:
            num3 = 3*num2 - num1
            den3 = 3*den2 - den1
            num1 = num2
            num2 = num3
            den1 = den2
            den2 = den3
            p = num2
            q = den2

        if i % 2 == 0:
            total = total - p/q
        else:
            total = total + p/q

    print(total)

while True:
    menu()
    pilih = input("Masukkan :")

    if pilih == "1":
        pangkat()
    elif pilih == "2":
        deret()
    elif pilih == "0":
        print("keluar")
        break
    else:
        print("pilihan tidak ada")
```
